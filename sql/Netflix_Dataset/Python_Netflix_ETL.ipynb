```python
#import sql package
from sqlalchemy import create_engine

engine = create_engine("postgresql+psycopg2://postgres:???@???:???/netflix_db")

engine.connect()
```

connection settled!
Syntax create_engine("postgresql+psycpg2://user:password@host:server/table_name")

```python
#check connection
pd.read_sql("SELECT 100;", engine)
```

If tested number appear, it works then.

```python
#import dataset to sql
df.to_sql('netflix_raw', con = engine, index = False, if_exists = 'append')
```

```python
#check imported dataset
pd.read_sql("SELECT COUNT(*) FROM netflix_raw;", engine)
```
